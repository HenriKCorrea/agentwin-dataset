```python
# 03_explore_scope.ipynb
import geopandas as gpd
import folium, json

dcas = json.load(open("datasets/raw/dcas_ev_stations_2026-07-22.json"))["data"]
osm_gdf = gpd.read_file("datasets/raw/osm_substations_nyc_2026-07-22.geojson")

m = folium.Map(location=[40.71, -74.00], zoom_start=11)

# Plot DCAS stations by agency (colour-coded)
agency_colors = {
    "NYPD": "blue",
    "FDNY": "red",
    "DSNY": "green",
    "DOE": "orange",
    "NYCEM": "purple",
}
for row in dcas:
    folium.CircleMarker(
        location=[float(row["latitude"]), float(row["longitude"])],
        radius=4,
        color=agency_colors.get(row["agency"], "gray"),
        popup=f"{row['agency']} | {row['station_name']} | {row['type_of_charger']}",
    ).add_to(m)

# Plot OSM substations
for _, sub in osm_gdf.iterrows():
    centroid = sub.geometry.centroid
    folium.Marker(
        location=[centroid.y, centroid.x],
        icon=folium.Icon(color="black", icon="bolt", prefix="fa"),
        popup=sub.get("name", "Substation"),
    ).add_to(m)

m.save("notebooks/scope_map.html")
```
