```python
import pandas as pd
import geopandas as gpd
import folium

df = pd.read_csv("../data/raw/dcas_ev_stations.csv")
ev_stations = gpd.GeoDataFrame(
    df,
    geometry=gpd.points_from_xy(df["longitude"], df["latitude"]),
    crs="EPSG:4326",
)
substations = gpd.read_file("../data/raw/osm_substations.geojson")

map = folium.Map(location=[40.71, -74.00], zoom_start=11)

for _, ev in ev_stations.iterrows():
    folium.Marker(
        location=[ev.geometry.y, ev.geometry.x],
        icon=folium.Icon(color="green", icon="car", prefix="fa"),
        popup=f"{ev['agency']} | {ev['station_name']} | {ev['type_of_charger']}",
    ).add_to(map)

# Plot OSM substations
for _, sub in substations.iterrows():
    centroid = sub.geometry.centroid
    folium.Marker(
        location=[centroid.y, centroid.x],
        icon=folium.Icon(color="black", icon="bolt", prefix="fa"),
        popup=sub.get("name", "Substation"),
    ).add_to(map)

map.save("scope_map.html")
```
